Notice that we only got through August - that is actually correct because the NERIS website shows that data feeds are delayed by 60 days. Also, it runs through 8/2 and not 8/1 because the incidents are logged in UTC, not local time zone. Converting to EDT would make this last log 8/1/26 at 8:22 PM.
